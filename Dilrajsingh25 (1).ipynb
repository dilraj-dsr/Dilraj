```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

matches = pd.read_csv(r'C:\Users\dilra\OneDrive\Desktop\pythonproject\project\matches_sample.csv')
deliveries = pd.read_csv(r'C:\Users\dilra\OneDrive\Desktop\pythonproject\project\deliveries_sample.csv')
matches
deliveries 
```

The below bargraph process the top 10 batsman with highest runs in above datasets:
1. the first step is to groupby batsman categories and their run 
2. sum the values of run
3. sort values in descending order 
5. use head for first 10 values.

```python
top_runs = deliveries.groupby("batsman")["batsman_runs"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_runs.plot(kind="bar")
colors = plt.cm.plasma(range(0, 256, int(256/len(top_runs))))

bars = plt.bar(top_runs.index, top_runs.values, color=colors)

plt.title("Top 10 Run Scorers in IPL")
plt.xlabel("Batsman")
plt.ylabel("Total Runs")
plt.show()
```

This plot shows the strike rate by an individual batsman maximum sr = 600


```python
balls_faced = deliveries[deliveries["batsman_runs"] >= 0].groupby("batsman").size()
runs = deliveries.groupby("batsman")["batsman_runs"].sum()

strike_rate = (runs / balls_faced) * 100
strike_rate = strike_rate.sort_values(ascending=False).head(10)


bars = plt.bar(strike_rate.index, strike_rate.values, color=colors)

strike_rate.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Strike Rates")
plt.ylabel("Strike Rate")
plt.show()
```

This plot shows the best and top wickettakers and arranged in descending orders and 0 values are avoided by notna

```python
wickets = deliveries[deliveries["dismissal_kind"].notna()]
top_wickets = wickets.groupby("bowler").size().sort_values(ascending=False).head(10)

top_wickets.plot(kind="bar", figsize=(10,5),color=colors)
plt.title("Top 10 Wicket Takers")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.show()
```

This plot shows the best economical bowler who have concedded less runs in a spell
the total value is divided by 6 to get economy

```python
economy = deliveries.groupby("bowler")["total_runs"].sum() / (deliveries.groupby("bowler").size() / 6)
economy = economy.sort_values().head(10)

economy.plot(kind="bar", figsize=(10,5),color=colors)
plt.title("Top 10 Economical Bowlers")
plt.ylabel("Economy Rate")
plt.show()
```

This info shows that how many matches are held per season in different years

```python
season_matches = matches.groupby("season")["id"].count()
season_wins = matches.groupby(["season", "winner"])["id"].count()

("Matches per Season:", season_matches)
```

This plot shows the venue where most of the matches are played

```python
venue_stats = matches["venue"].value_counts().head(10)

plt.figure(figsize=(10,5))
venue_stats.plot(kind="bar",color=colors)
plt.title("Top Venues with Most Matches")
plt.xlabel("Venue")
plt.ylabel("Matches Played")
plt.show()
```
